### 1.
**Write a prompt to get ChatGPT to generate a catchy Instagram bio for a college student who loves cricket and music, and explain why your prompt is likely to get a good result.**

---
**My Prompt:**
> *"Act as an expert Gen-Z social media manager. Write 3 short, catchy, and trendy Instagram bios for a college student who is obsessed with playing cricket and listening to indie music. Use emojis, keep each option strictly under 150 characters, and give it a cool, energetic vibe."*

**Why it works:**
This prompt is highly effective because it leverages **Role Assignment** (*"Act as an expert..."*), sets strict **Constraints** (*under 150 characters, exactly 3 options, use emojis*), and explicitly defines the **Tone/Vibe** (*cool, energetic*). This prevents the AI from generating something boring, robotic, or overly lengthy.

### 5.
**List 3 common mistakes people make when writing prompts for LLMs, and for each, rewrite a bad prompt into a better one using best practices discussed in class.**

---
1. **Mistake 1: Being Too Vague (Zero Constraints).**
   * **Bad Prompt:** *"Write about cars."*
   * **Better Prompt:** *"Write a 3-paragraph blog post comparing the battery life and highway charging speed of the Tesla Model 3 versus the Hyundai Ioniq 5."*
2. **Mistake 2: Missing Context & Persona.**
   * **Bad Prompt:** *"Explain quantum physics."*
   * **Better Prompt:** *"Act as a friendly 5th-grade science teacher. Explain the basic concept of quantum physics to a 10-year-old using a simple, easy-to-understand analogy about building blocks or toys."*
3. **Mistake 3: Leaving Output Formatting Open-Ended.**
   * **Bad Prompt:** *"Give me data on population."*
   * **Better Prompt:** *"Generate a Markdown table showing the top 5 most populated cities in India, including columns for City Name, Population in 2023, and State."*
