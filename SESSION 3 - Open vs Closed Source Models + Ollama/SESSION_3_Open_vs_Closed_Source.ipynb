### 2.
**Create a table comparing open source and closed source AI models across 5 points: cost, customization, community support, data privacy, and update frequency. Use real model names (such as Llama 2, GPT-4, Falcon, Gemini) in your comparison.**

---
| Feature | Open Source (e.g., Llama 2, Falcon, Mistral) | Closed Source (e.g., GPT-4, Claude 3, Gemini) |
| :--- | :--- | :--- |
| **Cost** | The model itself is entirely free to download, but the developer must pay for their own hardware and server hosting costs to run it. | The developer must pay per-token API fees to the vendor or a flat monthly subscription cost. |
| **Customization** | Absolute control; developers can physically modify the model weights and deeply fine-tune the core neural architecture. | Highly restricted; developers can only use basic prompt engineering or fine-tune via a black-box API. |
| **Community Support** | Massive open communities (HuggingFace, GitHub) building free tools, scripts, and sharing optimizations publicly. | Support is strictly limited to the official vendor's documentation and enterprise customer service teams. |
| **Data Privacy** | Extremely high; you can run the model entirely locally (offline) so that sensitive user data never leaves your internal network. | Lower privacy; you must transmit your users' sensitive data to external third-party cloud servers for processing. |
| **Update Frequency** | Highly frequent, chaotic, and rapid updates driven by millions of global developers experimenting organically. | Controlled, stable, periodic updates released entirely at the discretion of the corporate vendor. |

---
**Exact Command Used:**  
`ollama run llama2`

**Output Observed:**  
After executing the command in PowerShell, Ollama immediately initiated a download for the `llama2` model manifest and weights (approx 3.8 GB) with a visual progress bar. Once the download completed, it successfully verified the SHA256 checksums, loaded the model into local memory, and finally presented an interactive `>>>` prompt waiting for my chat input natively in the terminal.
